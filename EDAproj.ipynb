```python
#cleaning
import pandas as pd
df=pd.read_csv("/content/adult.csv")
df=df.replace(" ?",None)
print(df.isna().sum().sum())
print(df.duplicated().sum())
df=df.fillna({'Workclass':'Unknown','Occupation':'Not Available'})
#6
df['Age']=df['Age'].fillna(df['Age'].mean())
print(df)
```

```python
df['Hours per Week']=df['Hours per Week'].fillna(df['Hours per Week'].median())
print(df['Hours per Week'])
#8
df['Workclass']=df['Workclass'].fillna(df['Workclass'].mode())
print(df['Workclass'])
```

```python
df.isna().sum()
print(df['Native Country'].ffill()) #Forward fill
print(df['Native Country'].bfill()) #Backward fill
```

```python
# Detecting Outliers
Q1=df['Age'].quantile(0.25)
Q3=df['Age'].quantile(0.75)
IQR=Q3-Q1
lower=Q1-1.5*IQR
upper=Q3+1.5*IQR
print(lower)
print(upper)

#Identifying outliers
age_outlier=df[(df['Age']<lower)|(df['Age']>upper)]
print(len(age_outlier))
```

```python
Q1=df['EducationNum'].quantile(0.25)
Q3=df['EducationNum'].quantile(0.75)
IQR=Q3-Q1
lower=Q1-1.5*IQR
upper=Q3+1.5*IQR
print(lower)
print(upper)

#Identifying outliers
education_outlier=df[(df['EducationNum']<lower)|(df['EducationNum']>upper)]
print(len(education_outlier))
```

```python
Q1=df['Hours per Week'].quantile(0.25)
Q3=df['Hours per Week'].quantile(0.75)
IQR=Q3-Q1
lower=Q1-1.5*IQR
upper=Q3+1.5*IQR
print(lower)
print(upper)

#Identifying outliers
hours_outlier=df[(df['Hours per Week']<lower)|(df['Hours per Week']>upper)]
print(len(hours_outlier))
```

```python
# Z score
from scipy.stats import zscore
df['Age_zscore']=zscore(df['Age'])
print('age')
z_outliers=df[df['Age_zscore'].abs()>3]
print(len(z_outliers))
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv("adult.csv")


# 1. Histogram
plt.figure(figsize=(8,5))

plt.hist(df["Age"], bins=20)

plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Count")

plt.show()


# 2. Bar Chart
plt.figure(figsize=(10,5))

df["Workclass"].value_counts().head(8).plot(kind="bar")

plt.title("Top Workclasses")
plt.xlabel("Workclass")
plt.ylabel("Number of People")

plt.xticks(rotation=45)

plt.show()


# 3. Box Plot
plt.figure(figsize=(8,5))

sns.boxplot(data=df, x="Income", y="Age")

plt.title("Age Distribution by Income")
plt.xlabel("Income")
plt.ylabel("Age")
plt.show()




# 4. Scatter Plot
plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x="Age",
    y="Hours per Week"
)

plt.title("Age vs Hours per Week")
plt.xlabel("Age")
plt.ylabel("Hours per Week")
plt.show()
```
